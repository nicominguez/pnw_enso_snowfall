### Que dataset queremos?
Uno donde una instancia es una temporada en una estación meteorológica, luego ~45 temporadas por estación (1981-2025), que incluye:
- Variables identificativas: *'station_id'*, *'season'*, 'station_name', 'latitude', 'longitude', 'zone_name'.
- Variables categóricas: 'huc8'.
- Variables numéricas: 'elevation_m', 'T_AVG', 'PRCP', 'SNOW', 'enso'.
    - Variables completitud: 'n_T_AVG', 'n_PRCP', 'n_SNOW', 'n_enso'.
Luego, queremos instancias bien seleccionadas, es decir, estaciones con un rango de años largo e idealmente ningún dato faltante.
