La catidad de nieve acumulada (SWE) del PNW es un recurso hídrico clave ya que alimenta el riego y el consumo de la región en la primavera. La anomalía en temperatura del pacífico (ENSO) es el principal predictor exógeno estacional con meses de antelación. 

En la brevedad de los años que se comenzo a medir los efectos de El Niño en la zona del *PNW*, se ha observado una temperatura generalmente inferior a otros años y una precipitación inferior aunque con menos consistencia. Estos podrían explicar en parte la nieve. Por ejemplo, el 2015 fue un año muy caluroso y bastante seco:

Esto da a entender que el SWE no esta explicado completamente por los factores quizas mas intuitivos: temperatura y precipitacion, sino muchos otros que quedan fuera del scope de esta practica, entre ellos, el ENSO. Asumimos esta incertidumbre y nos centramos en el efecto exclusivo del ENSO. Controlarlas puede eliminar precisamente parte de la vía por la que ENSO se relaciona con SWE.
